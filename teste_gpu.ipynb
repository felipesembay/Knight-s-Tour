## Treinamento 6x6 (Colab GPU)

Passos: montar Drive (persistência) → clonar/atualizar repo → instalar deps → treinar.

```python
# Monta Google Drive p/ nao perder progresso se a sessao Colab cair
from google.colab import drive
drive.mount('/content/drive')

DRIVE_DIR = "/content/drive/MyDrive/Knight-s-Tour"
import os
os.makedirs(DRIVE_DIR, exist_ok=True)
```

```python
# Clona ou atualiza o repo
REPO_DIR = "/content/Knight-s-Tour"
if not os.path.exists(REPO_DIR):
    !git clone https://github.com/felipesembay/Knight-s-Tour {REPO_DIR}
else:
    !cd {REPO_DIR} && git pull

!pip install -q -r {REPO_DIR}/RL/requirements.txt
```

```python
# Redireciona models/ e logs/ para o Drive (sobrevive a queda de sessao)
import shutil

RL_DIR = f"{REPO_DIR}/RL"
for name in ["models", "logs"]:
    local_path = f"{RL_DIR}/{name}"
    drive_path = f"{DRIVE_DIR}/{name}"
    os.makedirs(drive_path, exist_ok=True)
    if os.path.islink(local_path):
        continue
    if os.path.isdir(local_path):
        shutil.rmtree(local_path)
    os.symlink(drive_path, local_path)

print("models ->", os.path.realpath(f"{RL_DIR}/models"))
print("logs   ->", os.path.realpath(f"{RL_DIR}/logs"))
```
